# Basic EDA

```python
import pandas as pd

df = pd.read_csv('company.csv')

print("First 5 rows of the DataFrame:")
print(df.head())

print("\nDataFrame info:")
print(df.info())

print("\nColumn names in the DataFrame:")
print(df.columns.tolist())
```

# Removing unwanted columns

```python
df.drop(columns=["EmployeeID", "Name"], inplace=True)

print("\nDataFrame after dropping 'EmployeeID' and 'Name' columns:")
print(df.head())
```

# Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Label encode the categorical columns
df["Gender"] = le.fit_transform(df["Gender"])
df["EducationLevel"] = le.fit_transform(df["EducationLevel"])

df = pd.get_dummies(df, columns=["Department", "City"], drop_first=True, dtype=int)

print("\nDataFrame after encoding categorical variables:")
print(df.head())
```

# Splitting in X and Y

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["Salary"])
Y = df["Salary"]

print("\nFeatures (X):")
print(X.head())

print("\nTarget variable (Y):")
print(Y.head())
```

# Getting training and testing sets

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print("\nShapes of training and testing sets:")
print(f"X_train: {X_train.shape}, Y_train: {Y_train.shape}")
print(f"X_test: {X_test.shape}, Y_test: {Y_test.shape}")
```

# Stardard Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nFirst 5 rows of the scaled training features:")
print(X_train_scaled[:5])
```

# Training Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_scaled, Y_train)

print("\n Weights and bias of the linear regression model:")
print("Weights:", model.coef_)
print("Bias:", model.intercept_)
```

# Testing Model

```python
y_pred = model.predict(X_test_scaled)

print("\nPredicted salaries for the test set:")
print(y_pred)
```

# Evaluating Model

```python
from sklearn.metrics import root_mean_squared_error, r2_score

rmse = root_mean_squared_error(Y_test, y_pred)
r2 = r2_score(Y_test, y_pred)

print("\nRoot Mean Squared Error:", rmse)
print("R-squared:", r2)
```
