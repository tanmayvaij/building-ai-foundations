```python
import pandas as pd
```

# Creating a Series

```python
# Creating a Series from a list
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)

# Creating a Series with custom index (labels)
index = ['A', 'B', 'C', 'D', 'E']
series_custom = pd.Series(data, index=index)
print(series_custom)
```

# Creating a Dataframe

```python
# Creating a sample DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df = pd.DataFrame(data)

# Displaying the DataFrame
print(df)

# Creating a DataFrame from a matrix (2D list)
matrix_data = [
    [1, 'Alice', 25],
    [2, 'Bob', 30],
    [3, 'Charlie', 35],
    [4, 'David', 40]
]

df_matrix = pd.DataFrame(matrix_data, columns=['ID', 'Name', 'Age'])
print(df_matrix)
```

# Updating Columns

```python
# Updating values in the DataFrame
df.loc[df["Name"] == "Aarav", "City"] = "Delhi"
print("DataFrame after updating Aarav's city:")
print(df)

# Updating values based on a condition
df.loc[(df["Marks"] < 70) & (df["Passed"] == True), "Passed"] = False
print("DataFrame after updating 'Passed' column based on Marks:")
print(df)

# Updating whole column
df["Department"] = df["Department"].replace("Computer", "CS")
print("DataFrame after updating 'Department' column:")
print(df)
```

# Deleting Columns , Rows

```python
# Deleting a column
df.drop(columns=["Phone"], inplace=True)
print("DataFrame after deleting 'Phone' column:")
print(df)

# Deleting multiple columns
df.drop(columns=["Passed", "City"], inplace=True)
print("DataFrame after deleting 'Passed' and 'City' columns:")
print(df)

# Deleting a row by index
df.drop(index=0, inplace=True)
print("DataFrame after deleting the first row:")
print(df)
```

# Sorting & Aggregation


```python
# sorting the DataFrame by a single column
df_sorted = df.sort_values(by="Marks", ascending=False)
print("DataFrame sorted by Marks in descending order:")
print(df_sorted)

# sorting the DataFrame by multiple columns
df_sorted_multi = df.sort_values(by=["Department", "Marks"], ascending=[True, False])
print("DataFrame sorted by Department (ascending) and Marks (descending):")
print(df_sorted_multi)

# Aggregation 

average_marks = df["Marks"].mean()
print("Average Marks:", average_marks)

max_marks = df["Marks"].max()
print("Maximum Marks:", max_marks)

min_marks = df["Marks"].min()
print("Minimum Marks:", min_marks)

```

# Grouping and Aggregation

```python
# Grouping and aggregation
grouped = df.groupby("Department")["Marks"].mean()
print("Average Marks by Department:")
print(grouped)

# Recreate the 'Passed' column before grouping
df["Passed"] = df["Marks"] >= 70

# Grouping by multiple columns and aggregating
grouped_multi = df.groupby(["Department", "Passed"])["Marks"].mean()
print("Average Marks by Department and Passed status:")
print(grouped_multi)
```

# Merging and Joining

```python
# Merging and joining DataFrames
df1 = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["Aarav", "Diya", "Vivaan"],
    "Marks": [85, 78, 67]
})

df2 = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "City": ["Delhi", "Pune", "Nashik", "Mumbai"]
})

# Inner join
merged_df = pd.merge(df1, df2, on="StudentID", how="inner")
print("Merged DataFrame:")
print(merged_df)

# Outer join
merged_df_outer = pd.merge(df1, df2, on="StudentID", how="outer")
print("Outer Merged DataFrame:")
print(merged_df_outer)

# Left join
merged_df_left = pd.merge(df1, df2, on="StudentID", how="left")
print("Left Merged DataFrame:") 
print(merged_df_left)

# Right join
merged_df_right = pd.merge(df1, df2, on="StudentID", how="right")
print("Right Merged DataFrame:")
print(merged_df_right)

# Cross join
cross_joined_df = pd.merge(df1, df2, how="cross")
print("Cross Joined DataFrame:")
print(cross_joined_df)


# Joining DataFrames using the join method
df1.join(df2.set_index("StudentID"), on="StudentID", how="inner")
```

# Concatinating

```python
# concatenating DataFrames
df3 = pd.DataFrame({
    "StudentID": [104, 105],
    "Name": ["Ananya", "Aditya"],
    "Marks": [91, 72]
})

# row-wise concatenation
concatenated_df = pd.concat([df1, df3])
print("Concatenated DataFrame:")
print(concatenated_df)

# column-wise concatenation
concatenated_df_col = pd.concat([df1, df3], axis=1)
print("Column-wise Concatenated DataFrame:")
print(concatenated_df_col)
```

# Pivot Table

```python
# pivoting the DataFrame
pivot_df = df.pivot_table(values="Marks", index="Department", columns="Passed", aggfunc="mean")
print("Pivot Table of average Marks by Department and Passed status:")
print(pivot_df)

# pivoting with multiple values
pivot_multi_df = df.pivot_table(values=["Marks", "Age"], index="Department", columns="Passed", aggfunc="mean")
print("Pivot Table of average Marks and Age by Department and Passed status:")
print(pivot_multi_df)
```

# Apply method

```python
# applying a custom function to the DataFrame
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"
    
df["Grade"] = df["Marks"].apply(grade)
print("DataFrame after applying custom function to create 'Grade' column:")
print(df)
```
